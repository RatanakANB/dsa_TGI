# RPL Dashboard - Exploratory Data Analysis (EDA)

**Dataset:** `RPL_Dashboard_V.1.csv`  
**Objective:** Perform data loading, data profiling, cleaning, feature engineering, and exploratory data analysis using `pandas` and `matplotlib`.

## 1. Environment Setup & Data Loading
Import required Python libraries (`pandas`, `numpy`, `matplotlib.pyplot`) and load the dataset into a Pandas DataFrame.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Configure pandas display options
pd.set_option('display.max_columns', None)

# Define file path and load CSV dataset
csv_file = 'RPL_Dashboard_V.1.csv'
df = pd.read_csv(csv_file)
print(f"Dataset successfully loaded!")
print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}")
```

## 2. Data Inspection & Profiling
Inspect the structure, column names, data types, missing values, and duplicate records in the dataset.

```python
# Display the first 5 records
df.head()
```

```python
# Display DataFrame metadata and column data types
df.info()
```

```python
# Check missing values per column
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df)) * 100
missing_df = pd.DataFrame({'Missing_Count': missing, 'Missing_Percentage (%)': missing_pct.round(2)})

print("--- Columns with Missing Values ---")
print(missing_df[missing_df['Missing_Count'] > 0])
print(f"\nTotal Duplicate Rows: {df.duplicated().sum()}")
```

## 3. Data Cleaning & Feature Engineering
- Clean text strings by stripping leading/trailing whitespace.
- Convert date attributes (`DOB`, `Date_Committee_Meeting`, `Date_Issue_EN`, `Date_Expiry_EN`) to `datetime` objects.
- Compute candidate **Age** at certification date.

```python
# Clean string columns
str_cols = df.select_dtypes(include=['object']).columns
for col in str_cols:
    df[col] = df[col].astype(str).str.strip()

# Convert date columns to datetime
date_cols = ['DOB', 'Date_Committee_Meeting', 'Date_Issue_EN', 'Date_Expiry_EN']
for col in date_cols:
    df[col + '_parsed'] = pd.to_datetime(df[col], format='%d-%b-%y', errors='coerce')

# Calculate Candidate Age (in years) at certification issue date
df['Age'] = (df['Date_Issue_EN_parsed'] - df['DOB_parsed']).dt.days // 365

# Display Age distribution summary
print("--- Candidate Age Summary Statistics ---")
df[['Age']].describe()
```

## 4. Demographic Analysis
Analyze the distribution of candidates by Gender and Age profile.

```python
# Gender breakdown (Count and Percentage)
gender_counts = df['Gender_EN'].value_counts()
gender_pct = df['Gender_EN'].value_counts(normalize=True) * 100

gender_summary = pd.DataFrame({
    'Candidate_Count': gender_counts,
    'Percentage (%)': gender_pct.round(2)
})
print("--- Candidate Breakdown by Gender ---")
gender_summary
```

```python
# Categorize candidates into Age Groups
bins = [0, 20, 30, 40, 50, 100]
labels = ['<20', '20-29', '30-39', '40-49', '50+']
df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels, right=False)

age_group_summary = df['Age_Group'].value_counts().sort_index()
age_group_df = pd.DataFrame({
    'Age_Group': age_group_summary.index,
    'Candidate_Count': age_group_summary.values,
    'Percentage (%)': (age_group_summary.values / len(df) * 100).round(2)
})
print("--- Candidate Distribution by Age Group ---")
age_group_df
```

## 5. Occupation & National Qualification Analysis
Explore top certified occupations, National Qualification (NQ) levels, and Assessment Testing (AT) centers.

```python
# Top 10 Occupations
top_occupations = df['Occupation_EN'].value_counts().head(10)
occ_df = pd.DataFrame({
    'Candidate_Count': top_occupations,
    'Percentage (%)': (top_occupations / len(df) * 100).round(2)
})
print("--- Top 10 Occupations (English) ---")
occ_df
```

```python
# National Qualification (NQ) Levels
nq_counts = df['NQ_EN'].value_counts()
nq_df = pd.DataFrame({
    'Candidate_Count': nq_counts,
    'Percentage (%)': (nq_counts / len(df) * 100).round(2)
})
print("--- National Qualification (NQ) Levels ---")
nq_df
```

```python
# Top 10 Assessment Centers (AT_EN)
at_counts = df['AT_EN'].value_counts().head(10)
at_df = pd.DataFrame({
    'Candidate_Count': at_counts,
    'Percentage (%)': (at_counts / len(df) * 100).round(2)
})
print("--- Top 10 Assessment Testing Centers ---")
at_df
```

## 6. Sponsorship & Timeline Analysis
Analyze candidate sponsors and certification trends over time.

```python
# Sponsor breakdown
sponsor_counts = df['Sponsors'].value_counts()
sponsor_df = pd.DataFrame({
    'Candidate_Count': sponsor_counts,
    'Percentage (%)': (sponsor_counts / len(df) * 100).round(2)
})
print("--- Certification Sponsors Breakdown ---")
sponsor_df
```

```python
# Annual trend of certification issue dates
df['Issue_Year'] = df['Date_Issue_EN_parsed'].dt.year
yearly_certs = df['Issue_Year'].value_counts().sort_index()

yearly_df = pd.DataFrame({
    'Year': yearly_certs.index.astype('Int64'),
    'Certificates_Issued': yearly_certs.values
})
print("--- Certificates Issued by Year ---")
yearly_df
```
